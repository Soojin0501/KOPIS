```python
import re
import pandas as pd

# 제거할 패턴 (문자열 끝에 오는 '등', '외', '外')
def clean_tail_words(item):
    return re.sub(r"(등|외|外)$", "", item).strip()

def split_and_clean(x):
    if pd.isna(x):
        return []
    items = [i.strip() for i in str(x).split(",") if i.strip()]

    clean_items = []
    for item in items:
        # 괄호 안 내용 제거
        item = re.sub(r"\(.*?\)", "", item).strip()
        # 끝에 붙은 '등', '외', '外' 제거
        item = clean_tail_words(item)
        if len(item) > 0:
            clean_items.append(item)
    return clean_items

```

```python
def standardize_names(name_list):
    """
    출연진/제작진 이름 리스트에서 괄호 안 내용 및 불필요한 문자 제거 후 표준화
    """
    if not isinstance(name_list, list):
        return []
    cleaned_names = []
    for name in name_list:
        # 괄호 안 내용 제거
        name = re.sub(r"\(.*?\)", "", name).strip()
        # 끝에 붙은 '등', '외', '外' 제거 (clean_tail_words 함수 재활용)
        name = clean_tail_words(name)
        if len(name) > 0:
            cleaned_names.append(name)
    return cleaned_names

# '출연진내용' 컬럼에 split_and_clean 함수를 적용하여 리스트 형태로 만든 후 표준화 (df 데이터프레임)
df['출연진내용_list'] = df['출연진내용'].apply(split_and_clean)
df['출연진내용_standardized'] = df['출연진내용_list'].apply(standardize_names)

# '출연진내용'과 '제작진내용' 컬럼에 split_and_clean 함수를 적용하여 리스트 형태로 만든 후 표준화 (df_complex 데이터프레임)
df_complex['출연진내용_list'] = df_complex['출연진내용'].apply(split_and_clean)
df_complex['제작진내용_list'] = df_complex['제작진내용'].apply(split_and_clean)

df_complex['출연진내용_standardized'] = df_complex['출연진내용_list'].apply(standardize_names)
df_complex['제작진내용_standardized'] = df_complex['제작진내용_list'].apply(standardize_names)

print("df 데이터프레임 출연진 표준화 결과:")
display(df[['출연진내용', '출연진내용_list', '출연진내용_standardized']].head())

print("\ndf_complex 데이터프레임 출연진 및 제작진 표준화 결과:")
display(df_complex[['출연진내용', '출연진내용_list', '출연진내용_standardized', '제작진내용', '제작진내용_list', '제작진내용_standardized']].head())
```

### 데이터셋 구축 및 GNN 모델 준비
- 출연진과 제작진 이름 표준화 -> GNN 모델 학습에 필요한 **그래프 데이터셋 구축**해야함

```python
# 1. 통합 노드 및 엣지 리스트 생성

# 인물/기관 노드 리스트: df와 df_complex의 출연진내용_standardized와 제작진내용_standardized 컬럼의 모든 이름을 모아서 중복을 제거합니다.
all_personnel_names = set()
for name_list in df['출연진내용_standardized']:
    all_personnel_names.update(name_list)
for name_list in df_complex['출연진내용_standardized']:
    all_personnel_names.update(name_list)
for name_list in df_complex['제작진내용_standardized']:
    all_personnel_names.update(name_list)

personnel_nodes = sorted(list(all_personnel_names))
print(f"총 인물/기관 노드 수: {len(personnel_nodes)}")
# print("인물/기관 노드 예시:", personnel_nodes[:10]) # 너무 많을 수 있으니 일부만 출력


# 공연 노드 리스트: 모든 공연의 고유 ID(index 또는 공연명)를 공연 노드로 사용합니다.
# 여기서는 데이터프레임의 index를 공연 ID로 사용하겠습니다.
# df와 df_complex의 인덱스를 합쳐서 사용
all_performance_indices = list(df.index) + list(df_complex.index)
performance_nodes = sorted(list(set(all_performance_indices))) # 혹시 모를 중복 제거 및 정렬
print(f"총 공연 노드 수: {len(performance_nodes)}")
# print("공연 노드 예시:", performance_nodes[:10])


# 장르 노드 리스트: df에서 '장르명'에 있는 모든 단일 장르를 장르 노드로 정의합니다.
genre_nodes = sorted(list(df['장르명'].unique()))
print(f"총 장르 노드 수: {len(genre_nodes)}")
print("장르 노드 리스트:", genre_nodes)


# 노드 타입별 ID 매핑 딕셔너리 생성
# GNN 라이브러리에서 노드를 정수 ID로 관리하기 때문에 매핑이 필요합니다.
# 인물/기관 노드는 0부터 시작, 공연 노드는 인물/기관 노드 수부터 시작, 장르 노드는 공연 노드 수 + 인물/기관 노드 수부터 시작
personnel_id_map = {name: i for i, name in enumerate(personnel_nodes)}
performance_id_map = {idx: i + len(personnel_nodes) for i, idx in enumerate(performance_nodes)}
genre_id_map = {genre: i + len(personnel_nodes) + len(performance_nodes) for i, genre in enumerate(genre_nodes)}

print("\n노드 ID 매핑 생성 완료.")
# print("인물/기관 ID 예시:", list(personnel_id_map.items())[:5])
# print("공연 ID 예시:", list(performance_id_map.items())[:5])
# print("장르 ID 예시:", list(genre_id_map.items())[:5])


# 2. 그래프 엣지 데이터 구축

# 참여 엣지(Participation Edge): 공연-인물/기관 연결
# df와 df_complex의 각 행을 순회하며 엣지를 기록합니다.
participation_edges = []

# df 데이터프레임 처리
for index, row in df.iterrows():
    performance_id = performance_id_map[index]
    for person_name in row['출연진내용_standardized']:
        if person_name in personnel_id_map: # 표준화 과정에서 빈 문자열이 남았을 경우 제외
             person_id = personnel_id_map[person_name]
             participation_edges.append((performance_id, person_id)) # (공연 노드 ID, 인물/기관 노드 ID) 엣지 추가

# df_complex 데이터프레임 처리
for index, row in df_complex.iterrows():
    performance_id = performance_id_map[index]
    for person_name in row['출연진내용_standardized']:
        if person_name in personnel_id_map:
            person_id = personnel_id_map[person_name]
            participation_edges.append((performance_id, person_id))

    for person_name in row['제작진내용_standardized']:
         if person_name in personnel_id_map:
            person_id = personnel_id_map[person_name]
            participation_edges.append((performance_id, person_id))


# 중복 엣지 제거 (하나의 인물/기관이 한 공연에 여러 역할로 참여해도 엣지는 하나만 필요)
participation_edges = list(set(participation_edges))

print(f"\n총 참여 엣지 수: {len(participation_edges)}")
# print("참여 엣지 예시:", participation_edges[:10])

# 엣지 데이터를 DataFrame으로 변환 (선택 사항이지만 관리가 용이)
participation_edges_df = pd.DataFrame(participation_edges, columns=['source', 'target'])
print("\n참여 엣지 데이터프레임 생성 완료.")
display(participation_edges_df.head())
```

```python
# 노드 피처 생성 (간단하게 모든 노드를 원-핫 인코딩처럼 표현)
# 실제 GNN에서는 더 의미있는 피처를 사용할 수 있습니다.
num_nodes = len(personnel_nodes) + len(performance_nodes) + len(genre_nodes)
# 모든 노드에 대한 단위 행렬을 피처로 사용 (가장 기본적인 형태)
x = torch.eye(num_nodes)

# 엣지 데이터 준비
# all_edges_df는 source, target 컬럼을 가진 DataFrame
# PyTorch Geometric은 엣지를 [2, num_edges] 형태의 텐서로 기대합니다.
# 첫 번째 행은 source 노드 인덱스, 두 번째 행은 target 노드 인덱스
edge_index = torch.tensor(all_edges_df.values.T, dtype=torch.long)


# 라벨 데이터 준비 (단일 장르 공연에 대한 라벨만 사용)
# PyTorch Geometric의 Data 객체는 노드 순서대로 라벨을 가집니다.
# 우리는 df의 index를 performance_id로 사용했고, 이는 performance_id_map에 매핑되어 있습니다.
# 학습에 사용할 공연 노드 (단일 장르 공연)의 ID와 해당 라벨을 매핑해야 합니다.

# 모든 공연 노드에 대한 초기 라벨을 -1 등으로 설정 (라벨이 없는 노드)
y = torch.full((len(performance_nodes),), fill_value=-1, dtype=torch.long) # 공연 노드 수 만큼 라벨 텐서 생성


# genre_labels 딕셔너리의 키를 df의 원래 인덱스로 사용
genre_labels = {}
# 장르명에 대한 클래스 인덱스 매핑 (0부터 len(genre_nodes)-1)
genre_class_map = {genre: i for i, genre in enumerate(genre_nodes)}

for index, row in df.iterrows():
    genre_name = row['장르명']
    # '대중음악'처럼 장르명에 괄호가 포함된 경우를 대비하여 통일
    clean_genre_name = genre_name.split('(')[0]

    if clean_genre_name in genre_class_map: # 변경: genre_id_map 대신 genre_class_map 사용
         # 변경: 장르 ID 대신 클래스 인덱스를 사용
         genre_class_index = genre_class_map[clean_genre_name]
         genre_labels[index] = genre_class_index # df의 원래 인덱스를 키로 사용, 클래스 인덱스를 값으로 저장


# 단일 장르 공연의 실제 라벨 채우기
for original_index, genre_class_index in genre_labels.items(): # 변경: genre_id 대신 genre_class_index 사용
    # original_index는 df의 인덱스입니다.
    # performance_id_map에서 해당 df 인덱스에 해당하는 전체 그래프 노드 ID를 찾습니다.
    performance_graph_id_full = performance_id_map[original_index]
    # y 텐서는 공연 노드만을 대상으로 하므로, 공연 노드의 0부터 시작하는 인덱스를 계산합니다.
    performance_graph_id_relative = performance_graph_id_full - len(personnel_nodes)
    y[performance_graph_id_relative] = genre_class_index # 변경: 장르 ID 대신 클래스 인덱스 저장


# GNN 학습은 라벨이 있는 단일 장르 공연 노드에 대해서만 수행
# 학습 마스크 생성: 라벨이 -1이 아닌 공연 노드만 True
train_mask = (y != -1)

# 복합 장르 공연 노드에 대한 마스크 생성 (예측 대상) - 기존 코드 유지
predict_mask = torch.zeros(len(performance_nodes), dtype=torch.bool)
for original_index in df_complex.index:
    performance_graph_id_full = performance_id_map[original_index]
    performance_graph_id_relative = performance_graph_id_full - len(personnel_nodes)
    predict_mask[performance_graph_id_relative] = True


# PyTorch Geometric Data 객체 생성
# x: Node features [num_nodes, num_node_features]
# edge_index: Graph connectivity in COO format with shape [2, num_edges]
# y: Node labels [num_nodes] - 여기서는 공연 노드에 대한 라벨만 (y 텐서의 크기는 공연 노드 수)
# train_mask: Mask for training nodes [num_nodes] - 여기서는 공연 노드에 대한 마스크 (train_mask의 크기는 공연 노드 수)

data = Data(x=x, edge_index=edge_index, y=y, train_mask=train_mask)

# Data 객체 확인
print("\nPyTorch Geometric Data 객체 생성 완료:")
print(data)
print(f"Number of nodes: {data.num_nodes}")
print(f"Number of edges: {data.num_edges}")
print(f"Number of training nodes (performances): {data.train_mask.sum()}")
print(f"Number of nodes with labels: {(data.y != -1).sum()}") # 라벨이 -1이 아닌 노드 수 (단일 장르 공연)

# 복합 장르 예측을 위한 마스크는 따로 저장
data.predict_mask = predict_mask
print(f"Number of prediction nodes (complex performances): {data.predict_mask.sum()}")
```

```python
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GCNConv

class GCN(nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super().__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, out_channels)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = F.dropout(x, p=0.5, training=self.training)
        x = self.conv2(x, edge_index)
        return F.log_softmax(x, dim=1)

# 모델 초기화
# data.num_node_features는 현재 113306 (원-핫 인코딩 피처 차원)
# 출력 채널 수는 장르 노드의 수 (8개)
model = GCN(in_channels=data.num_node_features, hidden_channels=64, out_channels=len(genre_nodes))

print("GCN 모델 정의 완료:")
print(model)
```

```python
import torch.optim as optim

# 옵티마이저 설정
# Adam 옵티마이저를 사용하며, 모델의 모든 파라미터를 최적화 대상으로 합니다.
optimizer = optim.Adam(model.parameters(), lr=0.01)

# 손실 함수 설정
# log_softmax 출력을 사용하므로 Negative Log Likelihood 손실 함수를 사용합니다.
criterion = nn.NLLLoss()

# 모델을 학습 모드로 설정
model.train()

# 학습 루프
epochs = 200 # 학습 횟수 설정 (필요에 따라 조정 가능)

for epoch in range(epochs):
    optimizer.zero_grad() # 그래디언트 초기화
    out = model(data.x, data.edge_index) # 모델 예측 수행

    # 학습 데이터에 해당하는 노드의 출력과 라벨만 사용하여 손실 계산
    # data.y는 공연 노드에 대한 라벨 텐서이고, train_mask는 학습에 사용할 공연 노드를 나타냅니다.
    # out 텐서의 크기는 전체 노드 수 x 장르 수 입니다.
    # 우리는 공연 노드 (personnel_nodes 수 이후부터)에 대한 출력만 필요합니다.
    # train_mask는 공연 노드 텐서 (y 텐서와 같은 크기)에 대한 마스크입니다.
    # 따라서 전체 출력(out)에서 공연 노드에 해당하는 부분만 슬라이싱한 후, train_mask를 적용합니다.

    performance_output = out[len(personnel_nodes) : len(personnel_nodes) + len(performance_nodes)]

    # train_mask가 True인 위치의 출력과 라벨에 대해서만 손실 계산
    loss = criterion(performance_output[data.train_mask], data.y[data.train_mask])

    loss.backward() # 역전파
    optimizer.step() # 옵티마이저 스텝

    if (epoch + 1) % 10 == 0:
        print(f'Epoch {epoch+1}/{epochs}, Loss: {loss.item():.4f}')

print("\n모델 학습 완료!")
```
